```python
# ============================================
# 🎬 КИНОГИД — VOILÀ + КНОПКИ + РУССКИЕ НАЗВАНИЯ
# ============================================
import pandas as pd, json, os, random
from datetime import datetime
from IPython.display import display, HTML, clear_output
import ipywidgets as widgets

# ══════════════════════════════════════════
# 1. ДАННЫЕ
# ══════════════════════════════════════════
if "movies" not in globals():
    for p in ("tmdb_5000_movies.csv", "data/tmdb_5000_movies.csv"):
        if os.path.exists(p):
            movies = pd.read_csv(p); break
    else:
        raise RuntimeError(f"❌ Нет tmdb_5000_movies.csv в папке: {os.getcwd()}")

if "genres_clean" not in movies.columns:
    movies["genres_clean"] = movies["genres"].apply(
        lambda s: " ".join(g["name"] for g in json.loads(s))
        if isinstance(s, str) else "")
if "year" not in movies.columns:
    movies["year"] = pd.to_datetime(movies["release_date"], errors="coerce").dt.year
movies = movies.dropna(subset=["runtime", "vote_average"]).copy()
movies["runtime"] = movies["runtime"].astype(int)

# ══════════════════════════════════════════
# 2. ПЕРЕВОД
# ══════════════════════════════════════════
try:
    from deep_translator import GoogleTranslator
    _translator = GoogleTranslator(source="en", target="ru")
    _translate_ok = True
except ImportError:
    _translate_ok = False

RU_TITLES_FILE = "ru_titles.json"
if os.path.exists(RU_TITLES_FILE):
    with open(RU_TITLES_FILE, encoding="utf-8") as f:
        _ru_cache = json.load(f)
else:
    _ru_cache = {}

def translate_title(title):
    if not title or not isinstance(title, str): return title
    if title in _ru_cache: return _ru_cache[title]
    if not _translate_ok: return title
    try:
        result = _translator.translate(title) or title
    except Exception:
        result = title
    _ru_cache[title] = result
    if len(_ru_cache) % 10 == 0:
        try:
            with open(RU_TITLES_FILE, "w", encoding="utf-8") as f:
                json.dump(_ru_cache, f, ensure_ascii=False, indent=2)
        except OSError:
            pass
    return result

# ══════════════════════════════════════════
# 3. ИСТОРИЯ
# ══════════════════════════════════════════
HISTORY_FILE = "history.json"

def load_history():
    if os.path.exists(HISTORY_FILE):
        try:
            with open(HISTORY_FILE, encoding="utf-8") as f:
                return json.load(f)
        except (json.JSONDecodeError, OSError):
            return []
    return []

def save_history(hist):
    try:
        with open(HISTORY_FILE, "w", encoding="utf-8") as f:
            json.dump(hist, f, ensure_ascii=False, indent=2)
    except OSError:
        pass

history = load_history()
watched_titles = [e["title"] for e in history]

# ══════════════════════════════════════════
# 4. ЖАНРЫ
# ══════════════════════════════════════════
GENRES = {
    "комедия":"Comedy","драма":"Drama","боевик":"Action",
    "фантастика":"Science Fiction","ужасы":"Horror","триллер":"Thriller",
    "мелодрама":"Romance","романтика":"Romance","приключения":"Adventure",
    "мультфильм":"Animation","анимация":"Animation","детектив":"Mystery",
    "фэнтези":"Fantasy","фентези":"Fantasy","документальный":"Documentary",
    "семейный":"Family","история":"History","военный":"War",
    "вестерн":"Western","музыка":"Music","криминал":"Crime",
}

def translate_genre_to_en(ru):
    ru = (ru or "").strip().lower()
    if ru in GENRES: return GENRES[ru]
    for k, v in GENRES.items():
        if k in ru: return v
    return ""

def genre_to_russian(s):
    if not isinstance(s, str): return "—"
    rev = {v.lower(): k for k, v in GENRES.items()}
    return ", ".join(rev.get(g.lower(), g) for g in s.replace("|", " ").split())

# ══════════════════════════════════════════
# 5. ПОИСК
# ══════════════════════════════════════════
def find_movies(genre_en, min_dur, max_dur, min_rating=6.0, count=5, exclude=None):
    exclude = exclude or []
    f = movies[movies['genres_clean'].str.contains(genre_en, case=False, na=False)]
    f = f[(f['runtime'] >= min_dur) & (f['runtime'] <= max_dur)]
    f = f[f['vote_average'] >= min_rating]
    f = f[~f['title'].isin(exclude)]
    return None if len(f) == 0 else f.sort_values('vote_average', ascending=False).head(count)

def random_movie():
    pool = movies[(~movies['title'].isin(watched_titles)) & (movies['vote_average'] >= 6.5)]
    return None if len(pool) == 0 else pool.sample(1).iloc[0]

# ══════════════════════════════════════════
# 6. ПУЗЫРИ
# ══════════════════════════════════════════
def bubble_bot(t, wide=False):
    max_w = "95%" if wide else "78%"
    return (f'<div style="display:flex; justify-content:flex-start; margin-bottom:10px;">'
            f'<div style="background:#fff; border:1.5px solid rgba(37,117,252,0.25); '
            f'border-radius:16px 16px 16px 4px; padding:12px 16px; font-size:15px; '
            f'color:#2d0a54; max-width:{max_w}; line-height:1.45;">{t}</div></div>')

def bubble_user(t):
    return (f'<div style="display:flex; justify-content:flex-end; margin-bottom:10px;">'
            f'<div style="background:linear-gradient(135deg,#2575fc,#6a11cb); '
            f'border-radius:16px 16px 4px 16px; padding:12px 16px; font-size:15px; '
            f'color:#fff; max-width:78%; line-height:1.45;">{t}</div></div>')

def movie_card(i, m):
    ru = translate_title(m['title'])
    return (f"<div style='line-height:1.55;'><div style='font-size:15px;font-weight:600;color:#2d0a54;'>"
            f"{i}. {ru}</div>"
            f"<div style='font-size:12px;color:#6a11cb;'>🎭 {genre_to_russian(m['genres_clean'])}</div>"
            f"<div style='font-size:12px;color:#2575fc;'>⏱️ {int(m['runtime'])} мин · ⭐ {m['vote_average']}/10</div></div>")

# ══════════════════════════════════════════
# 7. ИСТОРИЯ — ФУНКЦИИ
# ══════════════════════════════════════════
def yes_answer(text):
    return (text or "").strip().lower() in ("да","д","yes","y","конечно","ага","")

def add_to_history(user_name, movie_title, genre, rating, duration=0):
    entry = {"name": user_name, "title": movie_title, "genre": genre,
             "rating": rating, "duration": duration,
             "date": datetime.now().strftime("%d.%m.%Y %H:%M")}
    history.append(entry)
    if movie_title not in watched_titles:
        watched_titles.append(movie_title)
    save_history(history)
    return entry

def show_user_history(user_name):
    entries = [e for e in history if e["name"] == user_name]
    if not entries:
        return f"<b>{user_name}</b>, у тебя пока нет записей 📭"
    total = len(entries)
    avg = sum(e["rating"] for e in entries) / total
    best = max(entries, key=lambda e: e["rating"])
    fav = max(set(e["genre"] for e in entries),
              key=lambda g: sum(1 for e in entries if e["genre"] == g))
    rows = "".join(
        f"<div style='padding:4px 0; border-bottom:1px solid rgba(106,17,203,0.1);'>"
        f"🎥 <b>{e['title']}</b> — ⭐ {e['rating']}/10 "
        f"<span style='color:#6a11cb; font-size:11px;'>· {e['date']}</span></div>"
        for e in reversed(entries[-5:]))
    return (f"📊 <b>Твоя история, {user_name}:</b><br><br>"
            f"🎬 Просмотрено: <b>{total}</b><br>"
            f"⭐ Средняя оценка: <b>{avg:.1f}/10</b><br>"
            f"🏆 Лучший фильм: <b>{best['title']}</b><br>"
            f"🎭 Любимый жанр: <b>{fav.capitalize()}</b><br><br>"
            f"<div style='font-size:12px; color:#6a11cb;'>Последние записи:</div>{rows}")

def show_user_history_table(user_name):
    entries = [e for e in history if e["name"] == user_name]
    if not entries:
        return f"📋 <b>{user_name}</b>, пока нет записей"
    def pill(r):
        r = int(r)
        if r >= 9: return "background:linear-gradient(135deg,#6a11cb,#2575fc); color:#fff;"
        if r >= 7: return "background:#2575fc; color:#fff;"
        if r >= 5: return "background:#7dd3fc; color:#0c2c5b;"
        return "background:#ffd93b; color:#5a3d00;"
    rows = ""
    for i, e in enumerate(reversed(entries), 1):
        bg = "#f0f6ff" if i % 2 == 0 else "#ffffff"
        rows += (f"<tr style='background:{bg}; border-bottom:1px solid #e8e8f5;'>"
                 f"<td style='padding:4px;text-align:center;color:#2575fc;font-weight:600;'>{i}</td>"
                 f"<td style='padding:4px;color:#2d0a54;font-weight:600;'>{e['title']}</td>"
                 f"<td style='padding:4px;color:#6a11cb;font-size:11px;'>{e['genre']}</td>"
                 f"<td style='padding:4px;text-align:center;'><span style='display:inline-block;"
                 f"padding:2px 6px;border-radius:7px;font-size:11px;font-weight:700;{pill(e['rating'])}'>{e['rating']}</span></td>"
                 f"<td style='padding:4px;text-align:center;color:#5a5a8a;font-size:11px;'>{e['date'][:5]}</td></tr>")
    th = "padding:6px;background:linear-gradient(135deg,#4a0d8a,#2575fc);color:#fff;font-size:11px;text-align:center;"
    total = len(entries); avg = sum(e["rating"] for e in entries)/total
    return (f"📋 <b>История — {user_name}</b> · всего {total} · средняя {avg:.1f}<br>"
            f"<div style='max-height:250px;overflow-y:auto;border-radius:8px;margin-top:6px;'>"
            f"<table style='width:100%;border-collapse:collapse;font-family:Segoe UI,sans-serif;font-size:12px;'>"
            f"<thead><tr><th style='{th}'>№</th><th style='{th}'>Фильм</th>"
            f"<th style='{th}'>Жанр</th><th style='{th}'>⭐</th><th style='{th}'>Дата</th></tr></thead>"
            f"<tbody>{rows}</tbody></table></div>")

def user_achievements(user_name):
    entries = [e for e in history if e["name"] == user_name]
    if not entries: return "🏅 <b>Достижения</b> · пока пусто"
    count = len(entries); ratings = [e["rating"] for e in entries]
    genres = set(e["genre"].lower() for e in entries)
    badges = []
    if count >= 1:  badges.append(("🎬","Первый шаг"))
    if count >= 3:  badges.append(("🍿","Кинолюбитель"))
    if count >= 5:  badges.append(("🎥","Киноман"))
    if count >= 10: badges.append(("🏆","Легенда"))
    if max(ratings) == 10: badges.append(("⭐","Перфекционист"))
    if min(ratings) <= 3:  badges.append(("💔","Строгий критик"))
    if sum(ratings)/len(ratings) >= 8.5: badges.append(("🌟","Щедрый зритель"))
    if len(genres) >= 3: badges.append(("🎭","Жанровый гурман"))
    cards = "".join(
        f"<span style='display:inline-block;margin:3px 4px 3px 0;padding:6px 10px;"
        f"background:linear-gradient(135deg,#f3ecff,#e0f2ff);border-radius:10px;"
        f"border-left:3px solid #6a11cb;font-size:12px;'>{icon} {t}</span>"
        for icon, t in badges)
    return f"🏅 <b>Достижения ({len(badges)}):</b><br><br>{cards}"

def show_all_users_table():
    from collections import defaultdict
    if not history: return "📊 Пока никто не оценил ни одного фильма"
    users = defaultdict(lambda: {"c":0,"t":0,"g":set(),"l":""})
    for e in history:
        u = e["name"]; users[u]["c"] += 1; users[u]["t"] += e["rating"]
        users[u]["g"].add(e["genre"].lower())
        if e["date"] > users[u]["l"]: users[u]["l"] = e["date"]
    rows = sorted([{"n":n,"c":d["c"],"a":round(d["t"]/d["c"],1),"g":len(d["g"]),"l":d["l"]}
                   for n,d in users.items()], key=lambda r: (-r["c"], -r["a"]))
    th = "padding:6px;background:linear-gradient(135deg,#4a0d8a,#2575fc);color:#fff;font-size:11px;text-align:center;"
    body = ""
    for i, r in enumerate(rows):
        bg = "#f0f6ff" if i % 2 == 0 else "#ffffff"
        body += (f"<tr style='background:{bg};border-bottom:1px solid #e8e8f5;'>"
                 f"<td style='padding:6px;font-weight:600;color:#6a11cb;'>{r['n']}</td>"
                 f"<td style='padding:6px;text-align:center;'>{r['c']}</td>"
                 f"<td style='padding:6px;text-align:center;font-weight:600;color:#2575fc;'>{r['a']}</td>"
                 f"<td style='padding:6px;text-align:center;'>{r['g']}</td>"
                 f"<td style='padding:6px;text-align:center;font-size:11px;color:#5a5a8a;'>{r['l']}</td></tr>")
    return (f"📊 <b>Все зрители ({len(rows)}):</b><br><br>"
            f"<table style='width:100%;border-collapse:collapse;font-size:12px;'>"
            f"<thead><tr><th style='{th}'>Зритель</th><th style='{th}'>Фильмов</th>"
            f"<th style='{th}'>Средняя</th><th style='{th}'>Жанров</th><th style='{th}'>Последний</th></tr></thead>"
            f"<tbody>{body}</tbody></table>")

def show_full_history_table():
    if not history: return "📜 История пуста"
    rows = list(reversed(history))[-10:]
    th = "padding:5px;background:linear-gradient(135deg,#6a11cb,#2575fc);color:#fff;font-size:11px;text-align:center;"
    body = ""
    for i, e in enumerate(rows, 1):
        bg = "#f0f6ff" if i % 2 == 0 else "#ffffff"
        body += (f"<tr style='background:{bg};border-bottom:1px solid #e8e8f5;font-size:11px;'>"
                 f"<td style='padding:5px;text-align:center;color:#2575fc;font-weight:600;'>{i}</td>"
                 f"<td style='padding:5px;'>{e['name']}</td>"
                 f"<td style='padding:5px;font-weight:600;color:#2d0a54;'>{e['title']}</td>"
                 f"<td style='padding:5px;color:#6a11cb;'>{e['genre']}</td>"
                 f"<td style='padding:5px;text-align:center;font-weight:700;'>{e['rating']}</td>"
                 f"<td style='padding:5px;text-align:center;color:#5a5a8a;'>{e['date']}</td></tr>")
    return (f"📜 <b>Последние записи ({len(rows)} из {len(history)}):</b><br><br>"
            f"<table style='width:100%;border-collapse:collapse;'>"
            f"<thead><tr><th style='{th}'>№</th><th style='{th}'>Зритель</th>"
            f"<th style='{th}'>Фильм</th><th style='{th}'>Жанр</th><th style='{th}'>⭐</th><th style='{th}'>Дата</th></tr></thead>"
            f"<tbody>{body}</tbody></table>")

# ══════════════════════════════════════════
# 8. ЗОДИАК
# ══════════════════════════════════════════
ZODIAC_SIGNS = {
    "овен":{"emoji":"♈","genres":["боевик","приключения"],"trait":"энергия и страсть к действию"},
    "телец":{"emoji":"♉","genres":["драма","романтика"],"trait":"чувственность и любовь к красоте"},
    "близнецы":{"emoji":"♊","genres":["комедия","детектив"],"trait":"общительность и любознательность"},
    "рак":{"emoji":"♋","genres":["драма","семейный"],"trait":"эмоциональность и забота о близких"},
    "лев":{"emoji":"♌","genres":["приключения","боевик"],"trait":"лидерство и любовь к яркости"},
    "дева":{"emoji":"♍","genres":["детектив","история"],"trait":"аналитический ум и внимание к деталям"},
    "весы":{"emoji":"♎","genres":["романтика","комедия"],"trait":"гармония и стремление к красоте"},
    "скорпион":{"emoji":"♏","genres":["триллер","ужасы"],"trait":"страстность и тяга к тайнам"},
    "стрелец":{"emoji":"♐","genres":["приключения","комедия"],"trait":"свобода и любовь к путешествиям"},
    "козерог":{"emoji":"♑","genres":["драма","история"],"trait":"серьёзность и целеустремлённость"},
    "водолей":{"emoji":"♒","genres":["фантастика","фэнтези"],"trait":"новаторство и оригинальность"},
    "рыбы":{"emoji":"♓","genres":["мелодрама","фэнтези"],"trait":"мечтательность и богатое воображение"},
}

def pick_movie_for_genre(genre_ru, exclude=None):
    exclude = exclude or []
    genre_en = translate_genre_to_en(genre_ru) or "Comedy"
    pool = movies[
        (movies['genres_clean'].str.contains(genre_en, case=False, na=False)) &
        (movies['vote_average'] >= 6.5) &
        (~movies['title'].isin(exclude))
    ]
    return None if len(pool) == 0 else pool.sample(1).iloc[0]

def generate_horoscope(sign_key):
    sign = ZODIAC_SIGNS[sign_key]
    days = ["Пн","Вт","Ср","Чт","Пт","Сб","Вс"]
    used = list(watched_titles)
    movies_list = []
    rows = ""
    for day in days:
        genre = random.choice(sign["genres"])
        movie = pick_movie_for_genre(genre, exclude=used)
        if movie is not None:
            used.append(movie['title'])
            movies_list.append(movie)
            idx = len(movies_list)
            ru = translate_title(movie['title'])
            block = (f"<div style='font-size:12px;font-weight:700;color:#6a11cb;'>{idx}. {ru}</div>"
                     f"<div style='font-size:10px;color:#5a5a8a;'>⭐ {movie['vote_average']}/10 · "
                     f"{int(movie['runtime'])} мин · {genre}</div>")
        else:
            block = f"<div style='font-size:11px;color:#6a11cb;'>{genre} — нет</div>"
        rows += (f"<div style='display:flex;gap:8px;padding:6px 0;border-bottom:1px solid #f0f0f5;'>"
                 f"<div style='flex:0 0 32px;text-align:center;font-weight:700;color:#2575fc;font-size:12px;'>{day}</div>"
                 f"<div style='flex:1;'>{block}</div></div>")
    html = (f"🌟 <b>{sign['emoji']} Гороскоп для {sign_key.capitalize()}</b><br>"
            f"<div style='font-size:11px;color:#5a5a8a;margin:4px 0 8px;'>{sign['trait']}</div>"
            f"{rows}")
    return html, movies_list

# ══════════════════════════════════════════
# 9. UI
# ══════════════════════════════════════════
messages = []

chat_html = widgets.HTML(value="")
input_text = widgets.Text(
    placeholder="Введи имя и нажми Enter...",
    layout=widgets.Layout(width="380px")
)
send_btn = widgets.Button(
    description="➤ Отправить",
    button_style='warning',
    layout=widgets.Layout(width="130px", height="34px")
)
choices_output = widgets.Output()

def render_chat(subtitle="● онлайн"):
    body = "".join(messages)
    chat_html.value = f"""
    <div style="max-width:520px;margin:12px auto;padding:3px;border-radius:20px;
      background:linear-gradient(135deg,#2575fc,#6a11cb 50%,#8e2de2);
      box-shadow:0 10px 28px rgba(106,17,203,0.25);">
      <div style="background:#fdfdff;border-radius:17px;padding:20px;
        font-family:'Segoe UI',sans-serif;min-height:200px;max-height:520px;overflow-y:auto;">
        <div style="display:flex;align-items:center;gap:10px;padding-bottom:12px;
          border-bottom:1px solid rgba(106,17,203,0.15);margin-bottom:14px;">
          <div style="width:32px;height:32px;border-radius:50%;
            background:linear-gradient(135deg,#2575fc,#6a11cb);
            display:flex;align-items:center;justify-content:center;font-size:16px;">🤖</div>
          <div>
            <div style="font-size:14px;font-weight:600;color:#2d0a54;">Киногид</div>
            <div style="font-size:11px;color:#6a11cb;opacity:0.7;">{subtitle}</div>
          </div>
        </div>
        {body}
      </div>
    </div>
    """

def bot_says(t): messages.append(bubble_bot(t)); render_chat()
def bot_wide(t): messages.append(bubble_bot(t, wide=True)); render_chat()
def user_says(t): messages.append(bubble_user(t)); render_chat()

# ══════════════════════════════════════════
# 10. КНОПКИ
# ══════════════════════════════════════════
state = {"step": "name", "data": {}}

def clear_choices():
    with choices_output:
        clear_output(wait=True)

def show_buttons(options, callback, per_row=4, btn_min_width="70px"):
    """options: list of (label, value)"""
    clear_choices()
    btns = []
    for label, value in options:
        b = widgets.Button(
            description=label,
            layout=widgets.Layout(width='auto', min_width=btn_min_width, height='36px', margin='3px')
        )
        b.add_class('kg-btn')
        b.on_click(lambda _, cb=callback, v=value: _on_choice(cb, v))
        btns.append(b)
    rows = [widgets.HBox(btns[i:i+per_row],
                         layout=widgets.Layout(justify_content='center', flex_wrap='wrap'))
            for i in range(0, len(btns), per_row)]
    with choices_output:
        display(widgets.VBox(rows, layout=widgets.Layout(align_items='center')))

def _on_choice(callback, value):
    if state["step"] == "end": return
    clear_choices()
    input_text.disabled = True; send_btn.disabled = True
    try:
        callback(str(value))
    finally:
        if state["step"] == "end":
            input_text.disabled = True; send_btn.disabled = True
        else:
            input_text.disabled = False; send_btn.disabled = False

# Готовые наборы кнопок
YES_NO = [("✅ Да", "да"), ("❌ Нет", "нет")]
GENRE_BUTTONS = [
    ("🎭 Комедия", "комедия"), ("🎭 Драма", "драма"),
    ("🎭 Боевик", "боевик"), ("🎭 Фантастика", "фантастика"),
    ("🎭 Ужасы", "ужасы"), ("🎭 Триллер", "триллер"),
    ("🎭 Мелодрама", "мелодрама"), ("🎭 Приключения", "приключения"),
]
ZODIAC_BUTTONS = [(f"{v['emoji']} {k.capitalize()}", k) for k, v in ZODIAC_SIGNS.items()]
MENU_BUTTONS = [
    ("🌟 Гороскоп", "1"),
    ("🎲 Случайный", "2"),
    ("🎯 Подбор", "3"),
]
DURATION_BUTTONS = [
    ("⏱️ До 90", "1"), ("⏱️ 90–120", "2"),
    ("⏱️ 120–150", "3"), ("⏱️ Любая", "4"),
]
RATING_BUTTONS = [(str(i), str(i)) for i in range(1, 11)]

# ══════════════════════════════════════════
# 11. HANDLERS
# ══════════════════════════════════════════
def show_menu():
    bot_says("Что хочешь сделать?<br><br>"
             "<b style='color:#b8860b;'>1.</b> 🌟 Кино-гороскоп<br>"
             "<b style='color:#1e5cbf;'>2.</b> 🎲 Случайный фильм<br>"
             "<b style='color:#6a11cb;'>3.</b> 🎯 Подбор по жанру")
    state["step"] = "menu"
    show_buttons(MENU_BUTTONS, handle_menu, per_row=3, btn_min_width="130px")

def ask_restart():
    bot_says("🔄 <b>Начать заново?</b>")
    state["step"] = "restart"
    show_buttons(YES_NO, handle_restart, per_row=2, btn_min_width="110px")

def show_final_progress():
    if history:
        bot_wide(show_all_users_table())
        bot_wide(show_full_history_table())

def handle_name(text):
    name = text.strip().capitalize() or "дорогой зритель"
    state["data"]["name"] = name
    user_says(name)
    bot_says(f"Приятно познакомиться, <b>{name}</b>! ✨")
    show_menu()

def handle_menu(text):
    c = text.strip()
    if c not in ("1","2","3"):
        bot_says("Выбери 1, 2 или 3 🤔"); return
    user_says(c)
    state["data"]["menu"] = c
    if c == "1":
        bot_says(f"🌟 Какой у тебя знак зодиака?")
        state["step"] = "zodiac"
        show_buttons(ZODIAC_BUTTONS, handle_zodiac, per_row=4, btn_min_width="100px")
    elif c == "2":
        rand = random_movie()
        if rand is None:
            bot_says("😔 Все фильмы пересмотрены!")
            show_final_progress(); ask_restart(); return
        state["data"]["movie"] = rand
        state["data"]["genre_ru"] = genre_to_russian(rand['genres_clean']).split(",")[0].strip() or "комедия"
        messages.append(bubble_bot(movie_card("🎲", rand))); render_chat()
        bot_says("🍿 Судьба выбрала! 👀 Уже посмотрел(а)?")
        state["step"] = "watched"
        show_buttons(YES_NO, handle_watched, per_row=2, btn_min_width="110px")
    else:
        bot_says("🎭️ Какой жанр?")
        state["step"] = "genre"
        show_buttons(GENRE_BUTTONS, handle_genre, per_row=4, btn_min_width="120px")

def handle_zodiac(text):
    s = text.strip().lower()
    key = None
    for k in ZODIAC_SIGNS:
        if k in s or s in k: key = k; break
    if key is None:
        for k, v in ZODIAC_SIGNS.items():
            if v["emoji"] in s: key = k; break
    if key is None:
        key = random.choice(list(ZODIAC_SIGNS.keys()))
        bot_says(f"🤔 Такой знак не нашёл. Возьму <b>{key.capitalize()}</b>.")
    user_says(key.capitalize())
    html, mlist = generate_horoscope(key)
    if not mlist:
        bot_says("😔 Не удалось подобрать фильмы.")
        show_final_progress(); ask_restart(); return
    state["data"]["horoscope_movies"] = mlist
    bot_wide(html)
    bot_says(f"🎬 Какой фильм приглянулся?")
    state["step"] = "horo_pick"
    show_buttons([(str(i+1), str(i+1)) for i in range(len(mlist))],
                 handle_horo_pick, per_row=7, btn_min_width="50px")

def handle_horo_pick(text):
    mlist = state["data"]["horoscope_movies"]
    try:
        n = int(text.strip())
        if not (1 <= n <= len(mlist)): raise ValueError
    except:
        bot_says(f"Введи 1–{len(mlist)} 🤔"); return
    chosen = mlist[n-1]
    user_says(str(n))
    state["data"]["movie"] = chosen
    state["data"]["genre_ru"] = genre_to_russian(chosen['genres_clean']).split(",")[0].strip() or "комедия"
    messages.append(bubble_bot(movie_card("🎬", chosen))); render_chat()
    bot_says("👀 Уже посмотрел(а)?")
    state["step"] = "watched"
    show_buttons(YES_NO, handle_watched, per_row=2, btn_min_width="110px")

def handle_genre(text):
    g = text.strip().lower() or "комедия"
    state["data"]["genre_ru"] = g
    user_says(g.capitalize())
    bot_says("⏱️ Сколько времени?")
    state["step"] = "duration"
    show_buttons(DURATION_BUTTONS, handle_duration, per_row=4, btn_min_width="110px")

def handle_duration(text):
    c = text.strip()
    opts = {"1":(0,90,"до 90"),"2":(90,120,"90–120"),"3":(120,150,"120–150"),"4":(0,999,"любая")}
    if c not in opts: c = "4"
    min_d, max_d, dt = opts[c]
    user_says(dt)
    state["data"]["min_dur"] = min_d
    state["data"]["max_dur"] = max_d
    bot_says("🔎 Ищу... ⏳️")
    g_en = translate_genre_to_en(state["data"]["genre_ru"]) or "Comedy"
    s = find_movies(g_en, min_d, max_d, 6.0, 5, watched_titles)
    if s is None:
        bot_says("😔 Ничего не нашёл. Попробуем снова?")
        state["step"] = "retry_search"
        show_buttons(YES_NO, handle_retry_search, per_row=2, btn_min_width="110px")
        return
    state["data"]["suggestions"] = s
    bot_says(f"🎥 Вот {len(s)} вариантов:")
    for i, (_, m) in enumerate(s.iterrows(), 1):
        messages.append(bubble_bot(movie_card(i, m)))
    render_chat()
    bot_says(f"Выбери номер:")
    state["step"] = "pick"
    show_buttons([(str(i+1), str(i+1)) for i in range(len(s))],
                 handle_pick, per_row=5, btn_min_width="50px")

def handle_retry_search(text):
    if yes_answer(text):
        user_says("Да")
        bot_says("🎭️ Какой жанр?")
        state["step"] = "genre"
        show_buttons(GENRE_BUTTONS, handle_genre, per_row=4, btn_min_width="120px")
    else:
        user_says("Нет")
        show_final_progress(); ask_restart()

def handle_pick(text):
    s = state["data"]["suggestions"]
    try:
        n = int(text.strip())
        if not (1 <= n <= len(s)): raise ValueError
    except:
        bot_says(f"Введи 1–{len(s)} 🤔"); return
    chosen = s.iloc[n-1]
    user_says(str(n))
    state["data"]["movie"] = chosen
    bot_says(f"🎉 Твой выбор — <b>{translate_title(chosen['title'])}</b>! 🍿")
    bot_says("👀 Уже посмотрел(а)?")
    state["step"] = "watched"
    show_buttons(YES_NO, handle_watched, per_row=2, btn_min_width="110px")

def handle_watched(text):
    if yes_answer(text):
        user_says("Да")
        bot_says("⭐ Оцени фильм от 1 до 10:")
        state["step"] = "rating"
        show_buttons(RATING_BUTTONS, handle_rating, per_row=5, btn_min_width="50px")
    else:
        user_says("Нет")
        bot_says("😊 Ок! Возвращайся, когда посмотришь 🍿")
        show_final_progress(); ask_restart()

def handle_rating(text):
    try:
        n = int(text.strip())
        if not (1 <= n <= 10): raise ValueError
    except:
        bot_says("Нужно 1–10 🤔"); return
    user_says(f"{n}/10")
    name = state["data"]["name"]
    m = state["data"]["movie"]
    g = state["data"]["genre_ru"]
    ru_title = translate_title(m['title'])
    entry = add_to_history(name, ru_title, g, n, int(m.get('runtime', 0)))
    bot_says(
        "💾 Сохранено!<br><br>"
        f"👤 <b>{entry['name']}</b><br>🎥 {entry['title']}<br>"
        f"🎭 {entry['genre'].capitalize()}<br>⭐ {entry['rating']}/10<br>"
        f"📅 {entry['date']}"
    )
    bot_wide(show_user_history(name))
    bot_wide(show_user_history_table(name))
    bot_wide(user_achievements(name))
    show_final_progress()
    ask_restart()

def handle_restart(text):
    if yes_answer(text):
        user_says("Да")
        name = state["data"].get("name", "дорогой зритель")
        state["data"] = {"name": name}
        bot_says(f"🔄 <b>С возвращением, {name}!</b> ✨")
        show_menu()
    else:
        user_says("Нет")
        bot_says("👋 <b>До встречи!</b> Возвращайся за новыми кинооткрытиями 🍿")
        state["step"] = "end"
        clear_choices()

def process_input(text):
    handlers = {
        "name": handle_name, "menu": handle_menu, "zodiac": handle_zodiac,
        "horo_pick": handle_horo_pick, "genre": handle_genre,
        "duration": handle_duration, "retry_search": handle_retry_search,
        "pick": handle_pick, "watched": handle_watched,
        "rating": handle_rating, "restart": handle_restart,
    }
    if state["step"] in handlers:
        handlers[state["step"]](text)

# ══════════════════════════════════════════
# 12. ВВОД ЧЕРЕЗ ТЕКСТ (для имени)
# ══════════════════════════════════════════
def on_send(_):
    if state["step"] == "end": return
    text = input_text.value
    if not text.strip(): return
    input_text.value = ""
    input_text.disabled = True; send_btn.disabled = True
    try:
        process_input(text)
    finally:
        if state["step"] == "end":
            input_text.disabled = True; send_btn.disabled = True
        else:
            input_text.disabled = False; send_btn.disabled = False
            input_text.focus()

send_btn.on_click(on_send)
input_text.continuous_update = False
def _on_value_change(change):
    if change["name"] == "value" and change["new"].strip():
        on_send(None)
input_text.observe(_on_value_change, names="value")

# ══════════════════════════════════════════
# 13. СТАРТ
# ══════════════════════════════════════════
render_chat()
display(widgets.VBox([
    chat_html,
    widgets.HBox([input_text, send_btn], layout=widgets.Layout(justify_content="center")),
    choices_output
]))

bot_says("Привет! Я Киногид 🎬<br>Как тебя зовут? ✨")
state["step"] = "name"
```
